# 🎬 Movie Recommendation System

# Notebook 4: Feature Engineering

## Objective

The objective of this notebook is to prepare textual movie features for the recommendation engine by combining important attributes into a single feature called **tags**.

```python
import pandas as pd
import numpy as np

import re
```

```python
df = pd.read_csv("../data/processed/movies_cleaned.csv")

df.head()
```

```python
movies = df[
    [
        "title",
        "genres",
        "keywords",
        "cast",
        "director",
        "overview",
        "vote_average",
        "popularity"
    ]
]

movies.head()
```

```python
movies.isnull().sum()
```

```python
text_columns = [
    "genres",
    "keywords",
    "cast",
    "director",
    "overview"
]

movies[text_columns] = movies[text_columns].fillna("")
```

```python
movies["tags"] = (
    movies["genres"] + " " +
    movies["keywords"] + " " +
    movies["cast"] + " " +
    movies["director"] + " " +
    movies["overview"]
)
```

```python
movies = movies[
    [
        "title",
        "tags",
        "vote_average",
        "popularity"
    ]
]

movies.head()
```

```python
movies["tags"] = movies["tags"].str.lower()
```

```python
movies["tags"] = movies["tags"].apply(
    lambda x: re.sub(r"[^a-zA-Z0-9 ]", " ", x)
)
```

```python
movies["tags"] = movies["tags"].str.replace(
    r"\s+",
    " ",
    regex=True
).str.strip()
```

```python
movies.head()
```

```python
movies.to_csv(
    "../data/processed/movie_tags.csv",
    index=False
)

print("✅ Feature engineered dataset saved successfully!")
```

# Summary

Feature Engineering Completed

The following transformations were performed:

- Selected important recommendation features
- Filled missing textual values
- Combined multiple text columns into a single **tags** feature
- Converted text to lowercase
- Removed punctuation
- Removed unnecessary spaces

The processed dataset is now ready for vectorization and similarity calculation.
