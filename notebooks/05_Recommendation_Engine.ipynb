# 🎬 Movie Recommendation System

# Notebook 5: Recommendation Engine

## Objective

Build a Content-Based Movie Recommendation System using:

- CountVectorizer
- Cosine Similarity

The system will recommend the Top 10 most similar movies based on movie metadata.

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import pickle
```

```python
movies = pd.read_csv("../data/processed/movie_tags.csv")

movies.head()
```

```python
movies.shape
```

```python
cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors = cv.fit_transform(movies["tags"]).toarray()
```

```python
vectors.shape
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
def recommend(movie):

    movie = movie.lower()

    movie_index = movies[
        movies["title"].str.lower() == movie
    ].index

    if len(movie_index) == 0:
        print("Movie not found!")
        return

    movie_index = movie_index[0]

    distances = similarity[movie_index]

    movie_list = sorted(
        enumerate(distances),
        reverse=True,
        key=lambda x: x[1]
    )[1:11]

    print(f"\nTop 10 Recommendations for '{movies.iloc[movie_index].title}'\n")

    for i, movie_data in enumerate(movie_list, start=1):

        print(
            f"{i}. {movies.iloc[movie_data[0]].title}"
        )
```

```python
recommend("Avatar")
```

```python
recommend("Inception")
```

```python
recommend("The Dark Knight")
```

```python
recommend("Interstellar")
```

```python
pickle.dump(
    movies,
    open("../models/movies.pkl", "wb")
)

pickle.dump(
    similarity,
    open("../models/similarity.pkl", "wb")
)

print("✅ Model Saved Successfully!")
```
