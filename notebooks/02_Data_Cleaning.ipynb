# Data Cleaning

## Objective

The objective of this notebook is to clean the raw movie dataset by handling missing values, removing unnecessary columns, correcting data types, and preparing a clean dataset for feature engineering.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
df = pd.read_csv("../data/raw/movies.csv")

df.shape
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

missing.sort_values(ascending=False)
```

```python
columns_to_drop = [
    "homepage",
    "index"
]

df.drop(columns=columns_to_drop, inplace=True)
```

```python
df.isnull().sum()
```

```python
important_columns = [
    "genres",
    "keywords",
    "overview",
    "cast",
    "director"
]

df.dropna(subset=important_columns, inplace=True)
```

```python
print("Duplicates Before :", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates After :", df.duplicated().sum())
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.info()
```

```python
df.to_csv("../data/processed/movies_cleaned.csv", index=False)

print("✅ Cleaned dataset saved successfully!")
```

# Summary

The dataset has been cleaned successfully.

Cleaning steps performed:

- Removed unnecessary columns
- Handled missing values
- Removed duplicate records
- Reset index
- Saved cleaned dataset

The cleaned dataset is now ready for feature engineering.
